```python
import sqlite3
import pandas as pd 

 
```

```python
con = sqlite3.connect("ipl-cricket-analytics/data/raw/ipl.db")
```

```python
def query(sql):
    return pd.read_sql(sql, con)
```

```python
query ("SELECT COUNT (*) AS total_matches FROM matches; ")
```

```python
query('''
SELECT match_id, match_date, venue, match_winner
FROM matches
WHERE season = 2018
'''
)
```

```python
query('''
SELECT match_id, venue, city
FROM matches
WHERE city='Mumbai';
''' )
```

```python
query('''
SELECT match_id,season,match_winner,win_by_runs
FROM matches
WHERE win_by_runs>100;
''')
```

```python
query('''
SELECT match_id,season,team1,team2
FROM matches
WHERE match_winner IS NULL;
''')
```
